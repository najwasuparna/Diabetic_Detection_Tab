# IMPORT LIBRARY

```python
import pandas as pd
from google.colab import drive
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

# Mount Google Drive

```python
drive.mount('/content/drive')
```


# Load dataset dari Drive

```python
df = pd.read_csv("/content/drive/MyDrive/utsMesin_najwa_16039/diabetes_prediction_dataset.csv")
```

# Tampilkan 5 dataset teratas

```python
print("Jumlah data:", df.shape)
display(df.head())
```

SAMPLING DATA (AMBIL SEBAGIAN DATA)

```python
df = df.sample(n=2000, random_state=42)
```

ENCODING DATA KATEGORIKAL

```python
le = LabelEncoder()
df['gender'] = le.fit_transform(df['gender'])
df['smoking_history'] = le.fit_transform(df['smoking_history'])
```

FITUR & TARGET

```python
X = df.drop("diabetes", axis=1)
y = df["diabetes"]
```

MODEL YANG DIBANDINGKAN

```python
models = {
    "KNN": KNeighborsClassifier(),
    "Decision Tree": DecisionTreeClassifier(),
    "Logistic Regression": LogisticRegression(max_iter=5000)
}
```

TAMPILAN MODEL

```python
print("\n=== MODEL YANG DIGUNAKAN ===")
for m in models:
    print(m)
```

SPLIT DATA

```python
splits = [
    (0.2, "80:20"),
    (0.3, "70:30"),
    (0.1, "90:10")
]

results = []
```

TRAINING & EVALUASI

```python
for test_size, label in splits:
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=42
    )

    for name, model in models.items():
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        acc = accuracy_score(y_test, y_pred)

        results.append([name, label, acc])
```

TAMPILAN SPLIT DATA


```python
from sklearn.metrics import classification_report

for test_size, label in splits:
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=42
    )

    print(f"\n=== SPLIT {label} ===")
    print("Train:", X_train.shape)
    print("Test :", X_test.shape)

    for name, model in models.items():
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        acc = accuracy_score(y_test, y_pred)

        print(f"\nModel: {name} | Split: {label}")
        print("Accuracy:", acc)
        print(classification_report(y_test, y_pred))
```

EVALUASI

1. Logistic Regression

```python
from sklearn.metrics import confusion_matrix

model_lr = models['Logistic Regression']
model_lr.fit(X_train, y_train)
y_pred_lr = model_lr.predict(X_test)

print("Logistic Regression")
print(confusion_matrix(y_test, y_pred_lr))
print(classification_report(y_test, y_pred_lr))
```

2. Decision Tree

```python
model_dt = models['Decision Tree']
model_dt.fit(X_train, y_train)
y_pred_dt = model_dt.predict(X_test)

print("Decision Tree")
print(confusion_matrix(y_test, y_pred_dt))
print(classification_report(y_test, y_pred_dt))
```

3. KNN

```python
model_knn = models['KNN']
model_knn.fit(X_train, y_train)
y_pred_knn = model_knn.predict(X_test)

print("KNN")
print(confusion_matrix(y_test, y_pred_knn))
print(classification_report(y_test, y_pred_knn))
```

TABEL HASIL

```python
result_df = pd.DataFrame(results, columns=["Model", "Split", "Accuracy"])

print("\n=== HASIL AKURASI ===")
display(result_df)
```

TABEL PERBANDINGAN

```python
pivot_table = result_df.pivot(index="Model", columns="Split", values="Accuracy")

print("\n=== TABEL PERBANDINGAN ===")
display(pivot_table)
```

GRADIO WEB APP

```python
import gradio as gr

# =========================
# MODEL FINAL
# =========================
model_final = LogisticRegression(max_iter=5000)
model_final.fit(X, y)

# =========================
# FUNGSI PREDIKSI
# =========================
def predict_diabetes(gender, age, hypertension, heart_disease, bmi, HbA1c, glucose, smoking):
    data = pd.DataFrame([{
        "gender": gender,
        "age": age,
        "hypertension": hypertension,
        "heart_disease": heart_disease,
        "smoking_history": smoking,
        "bmi": bmi,
        "HbA1c_level": HbA1c,
        "blood_glucose_level": glucose
    }])

    prediction = model_final.predict(data)[0]
    return "🟥 Diabetes" if prediction == 1 else "🟩 Tidak Diabetes"

# =========================
# GRADIO UI (DENGAN PETUNJUK)
# =========================
interface = gr.Interface(
    fn=predict_diabetes,
    inputs=[
        gr.Number(
            label="Gender",
            info="0 = Perempuan, 1 = Laki-laki",
            value=0
        ),
        gr.Number(
            label="Usia",
            info="Masukkan umur dengan angka",
            value=30
        ),
        gr.Number(
            label="Hipertensi",
            info="0 = Tidak, 1 = Ya",
            value=0
        ),
        gr.Number(
            label="Penyakit Jantung",
            info="0 = Tidak, 1 = Ya",
            value=0
        ),
        gr.Number(
            label="BMI (Ukuran berat badan terhadap tinggi badan)",
            info="Contoh: 18.5 - 35",
            value=25
        ),
        gr.Number(
            label="HbA1c Level (Rata-rata kadar gula darah selama ±3 bulan terakhir)",
            info="Contoh: 4 - 9",
            value=5.5
        ),
        gr.Number(
            label="Glukosa Darah",
            info="Contoh: 80 - 200",
            value=120
        ),
        gr.Number(
            label="Riwayat Merokok",
            info="0 = Tidak Pernah, 1 = Masa lalu, 2 = Perokok",
            value=0
        )
    ],
    outputs=gr.Textbox(label="Hasil Prediksi"),
    title="🩺 Prediksi Diabetes",
    description="Masukkan data pasien untuk memprediksi apakah terkena diabetes atau tidak."
)

interface.launch()
```
